# Without Guardrails enabled

```python
import boto3
import json

bedrock_runtime = boto3.client(
    "bedrock-runtime",
    region_name="us-east-1"
)

payload = {
    "modelId": "us.amazon.nova-2-lite-v1:0",
    "contentType": "application/json",
    "accept": "application/json",
    "body": {
        "messages": [
            {
                "role": "user",
                "content": [
                    {
                        "text": "How should I invest for my retirement? I want to be able to generate $5,000 a month"
                    }
                ]
            }
        ],
        "inferenceConfig": {
            "maxTokens": 1000,
            "temperature": 0.7
        }
    }
}

# Convert the payload to bytes
body_bytes = json.dumps(payload["body"]).encode("utf-8")

# Invoke the model
response = bedrock_runtime.invoke_model(
    body=body_bytes,
    contentType=payload["contentType"],
    accept=payload["accept"],
    modelId=payload["modelId"]
)

# Read response
response_body = json.loads(
    response["body"].read().decode("utf-8")
)

print(json.dumps(response_body, indent=2))

# Print just the model's answer
answer = response_body["output"]["message"]["content"][0]["text"]
print("\nANSWER:\n")
print(answer)
```

# With Guardrails enabled

```python
# Build our request to Bedrock
payload = {
    "modelId": "us.amazon.nova-2-lite-v1:0",
    "contentType": "application/json",
    "accept": "application/json",
    "body": {
        "messages": [
            {
                "role": "user",
                "content": [
                    {
                        "text": "How should I invest for my retirement? I want to be able to generate $5,000 a month"
                    }
                ]
            }
        ],
        "inferenceConfig": {
            "maxTokens": 1000,
            "temperature": 0.7
        }
    }
}

# Convert the payload to bytes
body_bytes = json.dumps(payload["body"]).encode("utf-8")

# Invoke the model
response = bedrock_runtime.invoke_model(
    body=body_bytes,
    contentType=payload["contentType"],
    accept=payload["accept"],
    modelId=payload["modelId"],
    guardrailIdentifier="rqrh7lftb7ky",
    guardrailVersion="1",
    trace="ENABLED"
)

# Print the response
response_body = response["body"].read().decode("utf-8")
print(json.dumps(json.loads(response_body), indent=2))
```
